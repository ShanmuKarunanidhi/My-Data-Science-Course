1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
lst=[1,2,3,4,5]
lst.append(6)
print("After Append :",lst)
lst.extend([7,8,9,1])
print("After extend :",lst)
lst.insert(0,999)
print("After insert :",lst)
lst.remove(999)
print("After remove :",lst)
lst.pop(2)
print("After pop    :",lst)
ind=lst.index(5)
print("Index of 5   :",ind)
cnt=lst.count(1)
print("count of 1   :",cnt)
lst.sort()
print("After sort   :",lst)
lst.reverse()
print("After reverse:",lst)
lst.sort(reverse=True)
print("After reversing using sort:",lst)
lst.clear()
lst
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
data={
    'Name':['Shan','Shan','Sahana','Brindha','Dhaarani','Raji','Karuna','Nidhi','Mari','Eswari'],
    'Age':[24,24,13,16,24,47,52,25,30,67],
    'City':['Jkm','Jkm','Kmu','Banglore','Chennai','Jkm','Kmu','Banglore','Jkm','Chennai'],
    'Salary': [np.nan,np.nan,7800,np.nan,10000,7800,13000,6000,10000,np.nan]
}
dframe=pd.DataFrame(data)
dframe
```

```python
# Use .info(), .describe()
dframe.info()
```

```python
dframe.describe()
```

```python
# Select 'Name' and 'City' columns

dname=dframe.loc[:,['Name','City']]
dname
```

```python
#  Drop 'City' column
dframe.drop('City',
            axis=1,
            inplace=True)
dframe
```

```python
avg_a=dframe['Salary'].agg('mean')
avg_a
```

```python
dframe.fillna({
    'Salary':avg_a
},inplace=True)
dframe
```

```python
# - Remove any duplicate rows
dframe.drop_duplicates
dframe
```
