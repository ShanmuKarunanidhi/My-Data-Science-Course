Discuss the scenarios where each scaling technique is preferred and why.
- Standardization : 
  Best for algorithms sensitive to variance and normally distributed features. Example: PCA requires standardized data to correctly capture variance.Algorithms assuming Gaussian distribution (Logistic Regression, Linear Regression, LDA, PCA, SVM, Neural Networks).Preserves outliers, stabilizes variance, works well when features have different units/scales

  
- Min-Max Scaling  : 
Ideal for neural networks and clustering, where bounded ranges improve convergence. Example: KNN distance calculations are more meaningful when features are scaled.Used for distance related algorithmsRescales values to a fixed range (usually [0,1].Distance-based models (KNN, k-means, Neural Networks with bounded inputs).Ensures all features contribute equally, avoids dominance of large-scale features.

# 3. Feature Engineering:

1.Lets combine Capital_gain and Capital_loss as Total_capital.As both indicates the capital amount, combining them will give us an consolidated amount of net capital.Instead of treating gains and losses separately, this captures overall investment outcome. People with positive net capital are more likely to fall into the higher income bracket.

```python
adult['Net_capital'] = adult['capital_gain'] + adult['capital_loss']
adult.head(1)
```

2.Lets find capital activity score that says how much a person can earn in week.I added 1 to the formula as "If hours per week is 0 , 1 can balance the answer from calulating value as infinity".It is defined as Sum of capital gains and losses relative to hours worked.Measures how actively someone generates capital relative to their work hours. Could be predictive for financial status.

```python
adult['capital_activity_score'] = (adult['capital_gain'] + adult['capital_loss']) / (adult['hours_per_week'] + 1)
adult.head(1)
```

Lets check for skewness in data

```python
adult['capital_gain'].skew()
```

```python
adult['capital_loss'].skew()
```

Both capital loss and capital gain are right skewed.Log Transformation: log(x + 1) is common to handle zeros. It compresses large values while spreading out smaller ones.Since data have zeros in capital_gain and capital_loss, log transformation with a +1 shift is usually the best choice.capital_gain and capital_loss are highly right-skewed (many zeros, few large values).
Log transformation reduces the impact of extreme values and helps models like linear regression, logistic regression, or tree-based models perform better.Using np.log1p(x) ensures that zeros are handled safely (log(0+1)=0).

```python
adult['capital_gain_log'] = np.log1p(adult['capital_gain'])
adult['capital_loss_log'] = np.log1p(adult['capital_loss'])
adult
```
