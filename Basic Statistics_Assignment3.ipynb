### Descriptive Analytics for Numerical Columns
#### <u>Objective</u>: 
To compute and analyze basic statistical measures for numerical columns in the dataset

```python
# loadig dataset

import pandas as pd
sales_discount=pd.read_csv("C:/Users/shanm/Python_Practice/Datasets/sales_data_with_discounts.csv")
sales_discount.head()
```

#### a.Mean
##### Numerical columns --- Volume | Avg Price | Total Sales Value | Discount Rate (%) | Discount Amount | Net Sales Value

```python
# Volume
volume_mean=sales_discount['Volume'].mean()
print("volume_mean            :",volume_mean)

# Avg Price
AvgPrice_mean=sales_discount['Avg Price'].mean()
print("AvgPrice_mean          :",AvgPrice_mean)

# Total Sales Value
Total_Sales_Value_mean=sales_discount['Total Sales Value'].mean()
print("Total Sales Value_mean :",Total_Sales_Value_mean)

# Discount Rate (%)
Discount_Rate_mean=sales_discount['Discount Rate (%)'].mean()
print("Discount Rate (%)_mean :",Discount_Rate_mean)

# Discount Amount
Discount_Amount_mean=sales_discount['Discount Amount'].mean()
print("Discount Amount_mean   :",Discount_Amount_mean)

# Net Sales Value
Net_Sales_Value_mean=sales_discount['Net Sales Value'].mean()
print("Net Sales Value_mean   :",Net_Sales_Value_mean)
```

##### <u>Inference from mean</u>  : 
<br> * On average, each transaction of sales invloves 5 units at price of 10453 rupees. 
<br> * Discount reduces the total sales value of Rs.33812 with an average discount of 15% by 3346 Rs, leaving a net values of sales as 30466 Rs<br/>

#### b.Median

```python
# Volume
volume_median=sales_discount['Volume'].median()
print("volume_median            :",volume_median)

# Avg Price
AvgPrice_median=sales_discount['Avg Price'].median()
print("AvgPrice_median          :",AvgPrice_median)

# Total Sales Value
Total_Sales_Value_median=sales_discount['Total Sales Value'].median()
print("Total Sales Value_median :",Total_Sales_Value_median)

# Discount Rate (%)
Discount_Rate_median=sales_discount['Discount Rate (%)'].median()
print("Discount Rate (%)_median :",Discount_Rate_median)

# Discount Amount
Discount_Amount_median=sales_discount['Discount Amount'].median()
print("Discount Amount_median   :",Discount_Amount_median)

# Net Sales Value
Net_Sales_Value_median=sales_discount['Net Sales Value'].median()
print("Net Sales Value_median   :",Net_Sales_Value_median)
```

##### <u>Inference with median</u> :
<br> * While the average sales seems high due to fewer large transactions, median values shows that typical sale is Rs.5700 with 4 units sold at rate of 1450 with discount given at 16%.i.e 988 leaving a net price of 4677. 
<br> * Since median is not as affected as mean through outliers, we can conclude by comparing mean and median that mean is pulled towards right outliers<br/>

#### c.Mode

```python
# Volume
volume_mode=sales_discount['Volume'].mode()
print("volume_mode            :",volume_mode)


# Avg Price
AvgPrice_mode=sales_discount['Avg Price'].mode()
print("AvgPrice_mode          :",AvgPrice_mode[0])


# Total Sales Value
Total_Sales_Value_mode=sales_discount['Total Sales Value'].mode()
print("Total Sales Value_mode :",Total_Sales_Value_mode[0])


# Discount Rate (%)
Discount_Rate_mode=sales_discount['Discount Rate (%)'].mode()
print("Discount Rate (%)_mode :",Discount_Rate_mode[0])


# Discount Amount
Discount_Amount_mode=sales_discount['Discount Amount'].mode()
print("Discount Amount_mode   :",Discount_Amount_mode[0])


# Net Sales Value
Net_Sales_Value_mode=sales_discount['Net Sales Value'].mode()
print("Net Sales Value_mode   :",Net_Sales_Value_mode[0])


##### Categorical
# Day
Day_mode=sales_discount['Day'].mode()
print("Day_mode               :",Day_mode[0])


# SKU
SKU_mode=sales_discount['SKU'].mode()
print("SKU_mode               :",SKU_mode[0])


# City
City_mode=sales_discount['City'].mode()
print("City_mode              :",City_mode[0])


# BU
BU_mode=sales_discount['BU'].mode()
print("BU_mode                :",BU_mode[0])


# Brand
Brand_mode=sales_discount['Brand'].mode()
print("Brand_mode             :",Brand_mode[0])


# Model
Model_mode=sales_discount['Model'].mode()
print("Model_mode             :",Model_mode[0])
```

##### <u> Inference from Mode</u>:

#### d.Standard Deviation

```python
# Volume
volume_std=sales_discount['Volume'].std()
print("volume_std            :",volume_std)


# Avg Price
AvgPrice_std=sales_discount['Avg Price'].std()
print("AvgPrice_std          :",AvgPrice_std)


# Total Sales Value
Total_Sales_Value_std=sales_discount['Total Sales Value'].std()
print("Total Sales Value_std :",Total_Sales_Value_std)


# Discount Rate (%)
Discount_Rate_std=sales_discount['Discount Rate (%)'].std()
print("Discount Rate (%)_std :",Discount_Rate_std)


# Discount Amount
Discount_Amount_std=sales_discount['Discount Amount'].std()
print("Discount Amount_std   :",Discount_Amount_std)


# Net Sales Value
Net_Sales_Value_std=sales_discount['Net Sales Value'].std()
print("Net Sales Value_std   :",Net_Sales_Value_std)
```

<u>Inference from Standard Deviation</u> :-

<br> * The SD of average price is larger than its mean, which means there is huge variablity in product prices due to different product categories like Brand,Model and differnce in Pricing strategies.Price of certail products is very high
<br> * The SD of total sales price is high, which means Sales values fluctuate heavily. Some transactions are very high-value, others are much lower. Revenue is unpredictable. 
<br> * The SD of total sales price is high, which means Sales values fluctuate heavily. Some transactions are very high-value, others are much lower. Revenue is unpredictable.
<br> * Discount % is consitent as SD of discount% is low. Discount % is arount 15% for almost all the sales
<br> * SD of Discount Amount is large.Due to variation in product prices and volume of transactions,discount amount varies as it depends on price of product and volume
<br> * SD of Net value is high. It suggests that there is huge variablity in transaction sizes.There exist some high value sales  and some low valued salesprice.

## Conclusion

- The business experiences high variability and unpredictability in both sales volume and revenue.
- Outliers (large transactions) distort averages, so median and mode give a more realistic picture of typical sales.
- Demand is inconsistent — some products and days (like Thursday in City C, Jeera brand, Babaji Cream model) show strong sales, but overall sales fluctuate heavily.
- Pricing strategy is diverse — some products are priced very high, leading to wide variation in average prices.
- Discounting is consistent in percentage terms, but the absolute discount amount varies with transaction size.
 <br><b>This suggests that the business should:</b><br>
 - Focus on stabilizing demand (reduce reliance on sporadic large transactions).
 - Use median-based metrics for planning, since mean is skewed by outliers.
 - Leverage insights on high-performing brands, models, and days to optimize promotions.
