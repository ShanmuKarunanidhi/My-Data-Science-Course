```python
import pandas as pd
import numpy as np
```

```python
patient_details=pd.read_csv("C:/Users/shanm/Python_Practice/Datasets/Patient_Data.csv")
patient_details
```

```python
# Task 1
patient_details.info()
```

```python
# Task 2
selected4_columns = patient_details.loc[:,['PatientID', 'Department', 'Doctor', 'BillAmount']]
selected4_columns
```

```python
#Task 3
dropped2_columns = patient_details.drop(['ReceptionistID','CheckInTime'],
                                       axis=1,
                                       inplace=True)
patient_details
```

```python
# Task 4.Use groupby to find total bill amount per department
group = patient_details.groupby('Department')
total_bill = group['BillAmount'].sum()
total_bill
```

```python
#Task  5.Remove duplicate patient records based on PatientID.
duplicateRemoved = patient_details.drop_duplicates(subset=['PatientID'],inplace=True)
patient_details
```

```python
# Task 6.Fill missing BillAmount values with the mean bill amount.
billMean = patient_details['BillAmount'].mean()
print("Mean Amount:", billMean)
patient_details.fillna({
    'BillAmount' : billMean
    },inplace=True)
patient_details
```

```python
# 7.	Merge the billing dataset with patient dataset on PatientID.
```

```python
billing_data = pd.read_csv("C:/Users/shanm/Python_Practice/Datasets/Billing_Data.csv")
billing_data
```

```python
merged = pd.merge(patient_details,billing_data, on="PatientID")
merged
```

```python
# 8.	Concatenate an additional DataFrame that contains new patients for the current week (row-wise)    

additional = {
    'PatientID' : [106,107],
    'Name' : ['Shanmu','Kozhi'],
    'Department' : ['Dermatology','Orthopedics'],
    'Doctor' : ['Dr. Rose','Dr. Lee'],
    'BillAmount' : [7000,1000],
    'InsuranceCovered' : [3000,0],
    'FinalAmount' : [4000,1000],
}
new_Patients = pd.DataFrame(additional,index=[6,7])
new_Patients
```

```python
patients_Current_week = pd.concat([merged,new_Patients] )
patients_Current_week
```

```python
# 9.Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).
billing_Cat = pd.concat([patients_Current_week["InsuranceCovered"], patients_Current_week["FinalAmount"]], axis=1)
billing_Cat
```
