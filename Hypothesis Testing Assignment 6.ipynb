## HYPOTHESIS TESTING

Data Provided:<br/>
●	The theoretical weekly operating cost model: W = $1,000 + $5X<br/>
●	Sample of 25 restaurants with a mean weekly cost of Rs. 3,050<br/>
●	Number of units produced in a week (X) follows a normal distribution with a mean (μ) of 600 units and a standard deviation (σ) of 25 units<
●	ˉxˉ = sample mean weekly cost (Rs. 3,050)<br/>
●	μ = theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units<br/>)
●	σ = 5*25 uni<br/>ts
●	n = sample size (25 restauran<br/>ts)
br/>


```python
# W = 1000 + 5*(600)
n = 25
sample_mean = 3050
pop_mean = 1000 + 5*(600)
sigma_pop = 5*25
```

Step 1

Null Hypothesis
- The mean weekly operating cost is equal to 4000 (Population mean = 4000)<br/>

Alternate Hypotheis
- The mean weekly operating cost is greater than 4000 (Population mean > 4000) ### Right tailed test

```python
# Step 2
alpha = 0.05
```

```python
# Step 3
n = 25
sample_mean = 3050
```

```python
# Step 4
import scipy.stats as stats
import numpy as np
z_score = (sample_mean - pop_mean) / (sigma_pop/(np.sqrt(n)))
print(f"The Z score value is :{z_score}")

Pvalue = stats.norm.cdf(z_score)
print("The Probablity of z score according to z table is",Pvalue)                                 
                                    
```

```python
# Step 5
if Pvalue < alpha:
    print("Reject the null hypothesis which states that 'The mean weekly operating cost is equal to 4000 (Population mean = 4000)'meaning ,Accept the alternate hypothesis -The mean weekly operating cost is greater than 4000 (Population mean > 4000)")
else:
    print("Failed to reject null hypothesis which states that The mean weekly operating cost is equal to 4000 (Population mean = 4000)")
```

Conclusion :-

As per the restaurent owner's claim , the mean weekly operating cost is not equal to 4000 and it is greater than 4000
