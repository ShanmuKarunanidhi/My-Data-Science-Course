<u>Scenario</u></br>
A manufacturer of print-heads for personal computers is interested in estimating the mean durability of their print-heads in terms of the number of characters printed before failure. To assess this, the manufacturer conducts a study on a small sample of print-heads due to the destructive nature of the testing process.



# Task a) Population standard deviation not known

```python
# Data

n=15
sample = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
```

```python
# Assignment is to build interval with 99% confidence

import scipy.stats as stats

# Finding mean  bbbbb/gfc/
sample_mean = sum(sample)/len(sample)

print(f"sample mean is {sample_mean}")
```

```python
# Lets find out sample standard deviation

import pandas as pd
series = pd.Series(sample)
print("Data type of series : ",type(series))

sample_std = series.std()
print("standard deviation : ",sample_std)
```

```python
# Finding Interval 

import numpy as np
import scipy.stats as stats
cint = stats.t.interval(0.99,loc=sample_mean, df=14, scale=sample_std/np.sqrt(n))
interval = np.round(cint,3)
interval
```

```python
# Task b) Population standard deviation  known
```

```python
pop_std = 0.2

confInt = stats.norm.interval(0.99,loc=sample_mean,scale=pop_std/np.sqrt(n))
confInt
```
