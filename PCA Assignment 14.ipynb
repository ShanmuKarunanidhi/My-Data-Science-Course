# Task 2: Dimensionality Reduction with PCA:

### Standardize

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.model_selection import train_test_split

X = wine.drop("Type" , axis=1)
y = wine["Type"]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled.shape
```

### Implement PCA 

```python
from sklearn.decomposition import PCA
```

Initially Im giving n_components =  number of features. Then I will find cumsum and decide number of features needed actually

```python
pca = PCA()
X_pca = pca.fit_transform(X_scaled)
X_pca.shape
```

```python
explained_Var = pca.explained_variance_ratio_
explained_Var
```

```python
np.cumsum(explained_Var)
```

This means that by keeping faeture 1 (PC1) to feature 9 (PC9) , we can preserve 95% of information

Scree Plot

```python
plt.plot(range(1, len(pca.explained_variance_ratio_) + 1),
         pca.explained_variance_ratio_, marker='o')
plt.xlabel('Number of Components')
plt.ylabel('Explained Variance Ratio')
plt.title('Scree Plot')
plt.show()
```

```python
cum_var = np.cumsum(pca.explained_variance_ratio_)

plt.plot(range(1, len(cum_var)+1), cum_var, marker='o')
plt.axhline(y=0.9, color='r', linestyle='--')  # 90% threshold
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.show()
```

### Now , the optimal components of PCA has been decided as 9.Lets transform the data now

```python
pca = PCA(n_components = 9)
X_pca = pca.fit_transform(X_scaled)
X_pca.shape
```

# Task 3: Clustering with Original Data:

```python
from sklearn.cluster import KMeans
```

###  Clustering raw data without PCA components

```python
kmeans = KMeans(n_clusters=3, random_state=42)
labels_original = kmeans.fit_predict(X_scaled)
```

```python
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels_original, cmap='viridis')
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Clustering (Original Data)")
plt.show()
```

```python
from sklearn.manifold import TSNE
```

```python
tmodel = TSNE(n_components=2)
```

```python
tcomps = tmodel.fit_transform(X)
tcomps_df = pd.DataFrame(tcomps , columns=['TC1','TC2'])
tcomps_df['Type'] = wine['Type']
tcomps_df.head()
```

```python
sns.relplot(tcomps_df , x='TC1' , y='TC2',
           kind='scatter' ,
           height=3,
           aspect=2,
           hue = 'Type')
plt.show()
```

As per scatter plot, the clusters are not so good and gets overlapped with one another

### Evaluation Metric

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score

sil_score_orig = silhouette_score(X_scaled, labels_original)
db_score_orig = davies_bouldin_score(X_scaled, labels_original)

print("Silhouette Score:", sil_score_orig)
print("Davies-Bouldin Index:", db_score_orig)
```

silhouette_score around 0 means Overlapping clusters and DB_Index greater than 1 meaning poor clusters

# Task 6: Conclusion and Insights

Key findings and insights 

- PCA reduced dimensionality while preserving most variance.
- Clustering on PCA data was better based on metrics.
- Visualization improved significantly with PCA
-
Practical Implications

Use PCA when:
- Dataset has many features
- Features are highly correlated
- Visualization is needed


Use original data when:
- All features are important
- Interpretability matters


Recommendations
- Apply PCA before clustering for high-dimensional datasets
- Skip PCA if dataset is small and already meaningful
- Always compare both approaches
